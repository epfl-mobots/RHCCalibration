# Nb used to explore the power distribution

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', 2000)
```

## Configuration

```python
abcs = ['abc02','abc03','abc04','abc23']
```

## Main code

```python
from pathlib import Path
from configparser import ConfigParser

heater_keys = [f"h{i:02d}" for i in range(10)]

rows = []
for board_id in abcs:
    cfg_path = Path("RHCConfigs") / (board_id.replace("abc", "rhc", 1) + ".cfg")
    cfg = ConfigParser()
    cfg.read(cfg_path)
    rows.append({"board_id": board_id, **{h: cfg.getfloat("MaxPwmPower", h) for h in heater_keys}})
df = pd.DataFrame(rows)
print(df)

all_powers = np.array([])
for idx, row in df.iterrows():
    row = row.drop(labels=['board_id'])
    all_powers = np.concatenate((all_powers, row.to_numpy()))

avg_power = round(all_powers.mean(), 3)
std_power = all_powers.std()

print(f"Average power: {avg_power}, std of power: {std_power}")
```
